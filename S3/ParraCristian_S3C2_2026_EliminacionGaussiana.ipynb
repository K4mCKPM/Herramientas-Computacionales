ELIMINACIÓN GAUSSIANA

```python
#Escriba un código de eliminación Gaussiana para resolver el sistema Ax=B.

import numpy as np
N=np.random.randint(3, 7)
Arreglo=(np.random.random((N,N))*10.0)-5.0
B=(np.random.random((N,1))*10.0)-5.0
Sol = np.zeros((N,1))

print(f"Dimensión de la matriz: {N}\n")
print(f"El arreglo A es:\n{Arreglo}\n")
print(f"El vector B es:\n{B}\n")

for i in range(N):
  B[i] = B[i] / Arreglo[i][i]
  Arreglo[i,:] = Arreglo[i,:] / Arreglo[i,i]

  for j in range(i+1, N):
    B[j] = B[j] - B[i] * Arreglo[j,i]
    Arreglo[j,:] = Arreglo[j,:] - Arreglo[i][:] * Arreglo[j,i]

for k in range(N):
  # Como comienzo desde el último valor creo una variable que inicia desde el último elemento
  w = -k-1
  Sol[w, 0] = B[w, 0] - np.sum(Arreglo[w,:] * Sol[:,0])

print(f"La solución obtenida es:\n{Sol}\n")

#compare sus resultados con el paquete de numpy:
#https://numpy.org/doc/stable/reference/generated/numpy.linalg.solve.html
numpy_sol = np.linalg.solve(Arreglo,B)
print(f"Mediante el método linalg de numpy se obtiene:\n{numpy_sol}\n")
```
